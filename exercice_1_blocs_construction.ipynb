# Exercice 1 - Découvrez les blocs de construction de l'Apprentissage par Renforcement

L'objectif de cet exercice est de se familiariser avec le cycle fondamental de l'apprentissage par renforcement :
**observation → action → récompense**

On utilisera l'environnement classique **CartPole-v1** de la bibliothèque Gymnasium.


## Étape 1 : Explorer l'environnement CartPole

On commence par créer l'environnement et examiner ses deux espaces fondamentaux :

- **L'espace d'observation** : ce que l'agent perçoit à chaque instant
- **L'espace d'action** : les choix disponibles pour l'agent


```python
import gymnasium as gym
```

```python
env = gym.make("CartPole-v1")

print("Espace d'observation :")
print(f"Type    : {env.observation_space}")
print(f"Exemple : {env.observation_space.sample()}")
print()
print("L'observation contient 4 valeurs continues :")
print("  - Position du chariot")
print("  - Vitesse du chariot")
print("  - Angle du poteau")
print("  - Vitesse angulaire du poteau")
```

```python
print("Espace d'action :")
print(f"Type    : {env.action_space}")
print(f"Exemple : {env.action_space.sample()}")
print()
print("L'agent dispose de 2 actions discrètes :")
print("  - 0 : pousser le chariot vers la gauche")
print("  - 1 : pousser le chariot vers la droite")
```

## Étape 2 : Simuler un agent aléatoire sur 10 épisodes

On fait interagir un agent avec l'environnement. L'agent est le plus simple possible : il choisit ses actions au hasard.

On mesure la **récompense totale** obtenue sur chaque épisode. Dans CartPole, chaque pas où le poteau reste debout rapporte +1.


```python
NB_EPISODES = 10

# On exécute 10 épisodes (10 parties)
for episode in range(NB_EPISODES):
    # Réinitialise l'environnement pour démarrer une nouvelle partie
    observation, info = env.reset()

    # Initialise le score total de l'épisode
    total_reward = 0

    # Indique que l'épisode n'est pas encore terminé
    terminated = False
    truncated = False

    # Continue tant que la partie n'est pas finie
    while not terminated and not truncated:
        # Choisit une action aléatoire parmi celles disponibles
        action = env.action_space.sample()

        # Exécute l'action et récupère le nouvel état, la récompense
        # et les indicateurs de fin d'épisode
        observation, reward, terminated, truncated, info = env.step(action)

        # Ajoute la récompense obtenue au score total
        total_reward += reward

    # Affiche le score total obtenu pendant l'épisode
    print(f"Épisode {episode + 1:2d} — récompense totale : {total_reward}")

# Ferme proprement l'environnement
env.close()
```
